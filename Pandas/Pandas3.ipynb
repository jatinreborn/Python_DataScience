# Column Tranformation in Pandas

```python
import pandas as pd
import numpy as np
```

### Row & Column Selection 

```python
sData=pd.read_csv('Subdata_pandas/student.csv')
sData
```

```python
# Select a specific column
sData[['firstName','marks']]
```

```python
# based on name select row
sData.loc[sData.firstName=="Rohit"]
sData.loc[sData.id==103]
sData.info()
sData.loc[sData.marks>=90]
sData.loc[:3]
sData.iloc[:5,1:4]
```

### Filter

```python
# Select student firstName, age and fees who have marks more than 90

sData[['firstName','age','fees']].loc[sData.marks>=90]
```

```python
# Select student age equal, above 20
sData['age']>=20
sData[sData['age']>=20]
sData_age_filter=sData[sData['age']>=20]
sData_age_filter
```

```python
#Finding mean
mean=np.mean(sData['fees'])
mean
```

```python
sData.where(sData['age']>=20, other="Minor")
```

```python
meanmarks=np.mean(sData['marks'])
meanmarks
sData.where(sData['marks']>=meanmarks, other="Pass")
```
