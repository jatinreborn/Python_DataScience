String Indexing & Slicing

```python
str="Darkchocolate"
```

```python
#indexing
print(str[2])
print(str[-len(str)]) #first element using -ve index
print(str[-1]) #last element of str using -ve index
```

```python
#Slicing
print(str[0:4]) #st_idx =0 end_idx=4 step=1(d)
print(str[4:]) #st_idx =4 end_inx=len(str) (d) step=1(d)

#print(str[0:4:-1])  not valid 

print(str[0:4:1])

print(str[::-1])

print(str[3::2])
```

String Functions

```python
s="he is a coder "
```

```python
#Concatenation

print( s+ "It's real.")

#length of a string

length= len(s)
print(length)

#ends with substring
print( s.endswith("er"))

#replace old to new

print(s.replace("he","I am"))

#capitalize 
print(s.capitalize())

#find 1st index position of a substring

print(s.find("e"))

#count number of substring
print("count-e: ", s.count("e"))

#Upper lower
print(f"Upper: {s.upper()} lower: {s.lower()}")
```
