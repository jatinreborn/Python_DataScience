```python
fruits=["apple","orange","cherry","apple"]
print(fruits)
```

```python
#Indexing & Slicing
print(fruits[1])
print(fruits[-3])

#Slicing->string_name[start:stop:step]
print(fruits[1:3])

rev_list=fruits[::-1]
print(rev_list)
```

```python
#Modifying List
print("Org",fruits)
#add the element end of the list
fruits.append("Mango")
print(fruits)

#To find the length of list
len_fruits=len(fruits)
print("Length:",len_fruits)
#To remove the last element
fruits.pop()
print(fruits)
#to remove an element
"""fruits.remove("cherry")
print(fruits)
fruits[3]="Black"
print(fruits)"""

```

```python
nums=[4,2,5,3,6,1]
print(max(nums))  #max element from the list
print(min(nums))
nums.sort()
print(nums)
```

Looping

```python
student=["Raju",24,7.23,True]
#Using indexing 
for i in range(len(student)):
    print(student[i],end=" ")
#use access elements
for val in student:
    print(val,end=" ")
    
```
