Tuples are ordered, unchangeable and allows duplicate values

tuples are used to store multiple items in single variable

```python
#Type 1: Using Parenthesis
nums=(1,2,3)
mixed=(56,"gyo",45.76,True)
nested=(2,5,(5,6))
print(type(nums))

#Type 2: Without using Parenthesis
colors="red","green","blue"
print(colors,type(colors))


#Type 3: Using tuple construct

list=['a','b','c']
tuple_list=tuple(list)
print(tuple_list)

#Type 4: Single

value=(50,)
print(type(value))

```

Tuple Operation

```python
#Concatenation
tup1=(1,2)
tup2=(3,4)
tup3=tup1+tup2
print(tup3)

#Unchangeable

#tup2[1]=50
#print(tup2)

list_tuple=list(tup2)
print(tup2)

list_tuple[1]=50
tup2=tuple(list_tuple)
print(tup2)
```

```python
#Repetition 

mark=(40,)*5
print(mark)

tup=(10,20,30,40)
print(50 in tup)
```

Iterating over tuples

```python
color=("blue","black","green")
#for loop
for c in colors:
    print(c, end=" ")
print("\n")

#While 
i=0
while i<len(colors):
    print(color[i], end=" ")
    i+=1
```

Tuple Methods

```python
colors=("blue","green","yellow","red")
#count
print(colors.count("blue"))
#index
print(colors.index("red"))
```

Tuple Function

```python
nums=(32,45,23,76,89,97)
#lenght
print(len(nums))
#max and min
print("max= ",max(nums),"min= ",min(nums))
#sort
sortedNums=sorted(nums)
print(sortedNums)
```
