Dictionary Method

```python

#get
s_name=student.get('name')
print(s_name)

#key
k=student.keys() 
print(k)

#values
v=student.values()
print(v)

#Dict Items
d=student.items()
print(d)

#copy
c=student.copy()
print(c)

#default

dvalue=student.setdefault("pin",100123)
print(dvalue)
print(student)

#Update

student.update({"age":21})
print(student)

#pop

pop_value=student.pop('cgpa')
print(student)

#pop item - removes last key value from dictionary
student.popitem()
print(student)

student.clear()
```

Iteration over dictionary

```python
student={
    'id':"24CS057",
    'name':"Jatin",
    'age':20,
    'dept':"CSE",
    'cgpa':8.5
}

#print only keys from dict
for x in student:
    print(x)
#print only values from dict

for xv in student.values():
    print(xv)

#print key value pairs

for k,v in student.items():
    print("Key= ",k,"Val= ",v)
```
