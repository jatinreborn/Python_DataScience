```python
import numpy as np

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

years = np.array([2021, 2022, 2023, 2024])

# 1) Shape of sales data
print("Shape of sales data:", sales_data.shape)

# 2) First three restaurant data
print("\nFirst three restaurant records:\n", sales_data[:3])

# 3) Total sales per year
yearly_total = sales_data[:, 1:].sum(axis=0)
print("\nTotal sales per year:")
for yr, total in zip(years, yearly_total):
    print(f"{yr}: ₹{total:,}")

# 4) Maximum sales per restaurant
restaurant_max = sales_data[:, 1:].max(axis=1)
print("\nMaximum sales per restaurant:")
for restaurant_id, mx in zip(sales_data[:, 0], restaurant_max):
    print(f"Restaurant {restaurant_id}: ₹{mx:,}")

# 5) Minimum sales per year
yearly_min = sales_data[:, 1:].min(axis=0)
print("\nMinimum sales per year:")
for yr, min_val in zip(years, yearly_min):
    print(f"{yr}: ₹{min_val:,}")

# 6) Average sales per restaurant
avg_per_restaurant = sales_data[:, 1:].mean(axis=1)
print("\nAverage sales per restaurant:")
for restaurant_id, avg in zip(sales_data[:, 0], avg_per_restaurant):
    print(f"Restaurant {restaurant_id}: ₹{avg:,.2f}")

# 7) Average cumulative sales across all restaurants
avg_cumulative = sales_data[:, 1:].sum(axis=1).mean()
print(f"\nAverage cumulative sales across all restaurants: ₹{avg_cumulative:,.2f}")

# 8) Monthly average for every restaurant every year
print("\nMonthly average sales per restaurant per year:")
monthly_avg = sales_data[:, 1:] / 12
for restaurant_id, yearly_row in zip(sales_data[:, 0], monthly_avg):
    print(f"Restaurant {restaurant_id}:")
    for yr, monthly in zip(years, yearly_row):
        print(f"  {yr}: ₹{monthly:,.2f} per month")
```
