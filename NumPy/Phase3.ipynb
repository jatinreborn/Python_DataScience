```python
import numpy as np
import statistics as st
```

Agregate function

```python
array=np.array([10,20,30,40,50])
print("Sum:",np.sum(array))
print("Max:",np.max(array))
print("Min:",np.min(array))
print("count:",np.size(array))

#    d1  d2  d3  d4   d5
#due 10  20  30  40   50
#due 10  30  60  100  150

print("Cumulative sum:",np.cumsum(array))
print("Cumulative Product:",np.cumprod(array))
```

Statistical Function

```python
m1=np.array([22,23,20,20,21,20,17])
m2=np.array([20,15,25,50,30,35,20])
print("Mean of m1=",np.mean(m1),"\nMean of m2=",np.mean(m2),"\n")
print("Median of m1=",np.mean(m1),"\nMedian of m2=",np.median(m2),"\n")
print("mode of m1=",st.mode(m1),"\nMode of m2=",st.mode(m2),"\n")
print("Standard Diviation of m1:",np.std(m1),"\nStandard Diviation of m2:",np.std(m2))
```

```python
price=np.array([150,170,180,200,250,300])
sales=np.array([21,22,23,24,25,26])
print(np.corrcoef(price,sales))
```
